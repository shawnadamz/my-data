```python

%%writefile products.py

print("(a)______________________________________________________________________")
# Base class
class Product:
    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            print("Price cannot be negative.")

    def display_info(self):
        print("Product:", self.name)
        print("Price:", self.get_price(), "AED")


# Derived class 1
class FoodProduct(Product):
    def __init__(self, name, price, expiry_date):
        super().__init__(name, price)
        self.expiry_date = expiry_date

    def display_info(self):
        print("Food Product:", self.name)
        print("Price:", self.get_price(), "AED")
        print("Expiry Date:", self.expiry_date)


# Derived class 2
class ElectronicProduct(Product):
    def __init__(self, name, price, warranty):
        super().__init__(name, price)
        self.warranty = warranty

    def display_info(self):
        print("Electronic Product:", self.name)
        print("Price:", self.get_price(), "AED")
        print("Warranty:", self.warranty, "months")
```

The %%writefile products.py command saves the code in a separate file called products.py in Colab working environment.

```python
print("b________________________________________________________________________")
# Import the classes from our separate module
from products import Product, FoodProduct, ElectronicProduct

# Create three objects
product1 = Product("Notebook", 10)
product2 = FoodProduct("Milk", 8, "25-10-2026")
product3 = ElectronicProduct("Headphones", 150, 12)

# Display product information
print("SHOP MANAGEMENT SYSTEM")
print("----------------------")

product1.display_info()
print()

product2.display_info()
print()

product3.display_info()
```

```python
print("(c)_________________________________________________")
# Change the price of the milk
print("Original milk price:", product2.get_price(), "AED")

product2.set_price(9)

print("Updated milk price:", product2.get_price(), "AED")

# Try entering an invalid price
product2.set_price(-5)

print("Final milk price:", product2.get_price(), "AED")
```

Encapsulation is demonstrated by the __price attribute in the Product class. The double underscore restricts direct access through the usual attribute name. Instead, the get_price() and set_price() methods allow us to access and modify the price. The setter also prevents negative prices from being accepted.

Polymorphism is demonstrated by the display_info() method. Although the method has the same name in all three classes, it displays different information depending on the type of product.
